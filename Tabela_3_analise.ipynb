```python
todas = pd.read_excel(path, sheet_name=None, header=None)
print(todas.keys())
```

```python
bruto = pd.read_excel(path, header=None)
bruto
```

```python
dados = bruto.iloc[3:36].copy()
dados.columns = [
    "regiao",
    "superior_total",
    "superior_homens",
    "superior_mulheres",
    "cte_total",
    "cte_homens",
    "cte_mulheres",
    "esb_total",
    "esb_homens",
    "esb_mulheres",
]
dados = dados.reset_index(drop=True)
for col in dados.columns[1:]:
    dados[col] = pd.to_numeric(dados[col], errors="coerce")
dados
```

# 1. Homens e mulheres por região com curso superior

```python
superior = dados[["regiao", "superior_total", "superior_homens", "superior_mulheres"]].copy()
superior.columns = ["regiao", "total", "homens", "mulheres"]
superior
```

```python
superior.describe().round(2)
```
